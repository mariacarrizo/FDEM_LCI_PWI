```python
import numpy as np
import empymod as ep
import emg3d as e3
import matplotlib.pyplot as plt
from scipy.constants import mu_0
import math
import sys
sys.path.insert(1, 'src')
```

# how to measure the distorsion with topography?

```python
# HCP

res = [1e6, 10, 10] # resistivity model
depth = [0, -10]

freq = 9000
src_x = 0
src_y = 0
src_z = 0.1
rec_y = 0

offset =  2

alphas = np.linspace(0, np.pi /64)
alphas_deg = np.degrees(alphas)

HCP_ip = []
HCP_op = []

rec_x_ = []
rec_z_ = []

for alpha in alphas:

    rec_x = offset * np.cos(alpha)
    rec_x_.append(rec_x)
    rec_z = src_z + offset*np.sin(alpha)
    rec_z_.append(rec_z)
    
    src = [src_x, src_y, src_z, 0, 90+math.degrees(alpha)]
    rec = [rec_x, rec_y, rec_z, 0, 90+math.degrees(alpha)]
    
    Hs = ep.bipole(src, rec, depth, res, freqtime = freq, mrec=True, msrc=True, xdirect=None, verb=0 )*(2j * np.pi * freq * mu_0) 
    Hp = ep.bipole(src, rec, depth=[], res = [1e6], freqtime = freq, mrec=True, msrc=True, verb=0 )*(2j * np.pi * freq * mu_0) 
    
    HCP_op.append((Hs/Hp).imag.amp())
    HCP_ip.append((Hs/Hp).real.amp())
```

```python
fig, ax = plt.subplots(2,3, sharex=True, figsize = (10,4))

ax[0,0].plot(alphas_deg, np.array(HCP_op).ravel(), label = 'Q - Imag')
ax[0,1].plot(alphas_deg, np.array(HCP_ip).ravel(), label = 'IP - Real')
ax[0,2].plot(alphas_deg, rec_x_, 'g', label = 'rec x')
ax[1,2].plot(alphas_deg, rec_z_, 'g', label = 'rec z')

ax[1,0].plot(alphas_deg, np.abs(np.array(HCP_op).ravel()-HCP_op[0])/(HCP_op[0])*100, 'r', label='% Diff', )
ax[1,1].plot(alphas_deg, np.abs(np.array(HCP_ip).ravel()-HCP_ip[0])/(HCP_ip[0])*100, 'r', label='% Diff', )
fig.suptitle('HCP 2m offset')

ax[0,0].legend()
ax[0,1].legend()
ax[1,0].legend()
ax[1,1].legend()
ax[0,2].legend()
ax[1,2].legend()
ax[1,0].set_xlabel('alpha')
ax[1,1].set_xlabel('alpha')
ax[1,2].set_xlabel('alpha')

plt.tight_layout()
```

```python
sigmas = np.logspace(0,3)/1000

res = [1e6, 10, 10] # resistivity model
depth = [0, -10]

freq = 9000
src_x = 0
src_y = 0
src_z = 1
rec_y = 0

offset =  2

alphas_deg = np.arange(-5, 6, 0.5)
alphas = np.deg2rad(alphas_deg)


HCP_op = np.zeros((len(sigmas),len(alphas)))
HCP_ip = np.zeros((len(sigmas),len(alphas)))
VCP_op = np.zeros((len(sigmas),len(alphas)))
VCP_ip = np.zeros((len(sigmas),len(alphas)))
PRP_op = np.zeros((len(sigmas),len(alphas)))
PRP_ip = np.zeros((len(sigmas),len(alphas)))


for ir, s in enumerate(sigmas):
    for ia, alpha in enumerate(alphas):
        res = [1e6, 1/s, 1/s]

        rec_x = offset * np.cos(alpha)
        rec_xp = (offset+.1) * np.cos(alpha)
        rec_z = src_z + offset*np.sin(alpha)
        rec_zp = src_z + (offset+.1)*np.sin(alpha)
        if alpha < 0:
            assert rec_z > 0
            assert rec_zp > 0

        # HCP
        src = [src_x, src_y, src_z, 0, 90+math.degrees(alpha)]
        rec = [rec_x, rec_y, rec_z, 0, 90+math.degrees(alpha)]
        
        Hs = ep.bipole(src, rec, depth, res, freqtime = freq, mrec=True, msrc=True, xdirect=None, verb=0 )#*(2j * np.pi * freq * mu_0) 
        Hp = ep.bipole(src, rec, depth=[], res = [1e6], freqtime = freq, mrec=True, msrc=True, verb=0 )#*(2j * np.pi * freq * mu_0) 

        HCP_op[ir, ia] = (Hs/Hp).imag.amp()
        HCP_ip[ir, ia] = (Hs/Hp).real.amp()

        # VCP
        src = [src_x, src_y, src_z, 90, math.degrees(alpha)]
        rec = [rec_x, rec_y, rec_z, 90, math.degrees(alpha)]
        
        Hs = ep.bipole(src, rec, depth, res, freqtime = freq, mrec=True, msrc=True, xdirect=None, verb=0 )#*(2j * np.pi * freq * mu_0) 
        Hp = ep.bipole(src, rec, depth=[], res = [1e6], freqtime = freq, mrec=True, msrc=True, verb=0 )#*(2j * np.pi * freq * mu_0) 

        VCP_op[ir, ia] = (Hs/Hp).imag.amp()
        VCP_ip[ir, ia] = (Hs/Hp).real.amp()

        # PRP
        src = [src_x, src_y, src_z, 0, 90+math.degrees(alpha)]
        rec = [rec_xp, rec_y, rec_zp, 0, math.degrees(alpha)]
        rec_p = [rec_xp, rec_y, rec_zp, 0, 90+math.degrees(alpha)]
        
        Hs = ep.bipole(src, rec, depth, res, freqtime = freq, mrec=True, msrc=True, xdirect=None, verb=0 )#*(2j * np.pi * freq * mu_0) 
        Hp = ep.bipole(src, rec_p, depth=[], res = [1e6], freqtime = freq, mrec=True, msrc=True, verb=0 )#*(2j * np.pi * freq * mu_0) 
    
        PRP_op[ir, ia] = (Hs/Hp).imag.amp()
        PRP_ip[ir, ia] = (Hs/Hp).real.amp()

diff_HCP_op = np.zeros_like(HCP_op)
diff_HCP_ip = np.zeros_like(HCP_ip)
diff_VCP_op = np.zeros_like(VCP_op)
diff_VCP_ip = np.zeros_like(VCP_ip)
diff_PRP_op = np.zeros_like(PRP_op)
diff_PRP_ip = np.zeros_like(PRP_ip)

idx_min = np.argmin(np.abs(alphas_deg))

for l in range(len(sigmas)):
    diff_HCP_op[l,:] = (np.abs(HCP_op[l,:] - HCP_op[l,idx_min])/HCP_op[l,idx_min])*100
    diff_HCP_ip[l,:] = (np.abs(HCP_ip[l,:] - HCP_ip[l,idx_min])/HCP_ip[l,idx_min])*100
    diff_VCP_op[l,:] = (np.abs(VCP_op[l,:] - VCP_op[l,idx_min])/VCP_op[l,idx_min])*100
    diff_VCP_ip[l,:] = (np.abs(VCP_ip[l,:] - VCP_ip[l,idx_min])/VCP_ip[l,idx_min])*100
    diff_PRP_op[l,:] = (np.abs(PRP_op[l,:] - PRP_op[l,idx_min])/PRP_op[l,idx_min])*100
    diff_PRP_ip[l,:] = (np.abs(PRP_ip[l,:] - PRP_ip[l,idx_min])/PRP_ip[l,idx_min])*100
```
