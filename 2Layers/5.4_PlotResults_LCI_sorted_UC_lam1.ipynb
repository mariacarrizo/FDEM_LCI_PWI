```python
# Calculate Absolute error

abs_err_s1_c1_uc = np.abs(m_s1_c1_uc_g - mi_s1_c1_uc_g)
abs_err_s1_c2_uc = np.abs(m_s1_c2_uc_g - mi_s1_c2_uc_g)
abs_err_s1_c3_uc = np.abs(m_s1_c3_uc_g - mi_s1_c3_uc_g)
abs_err_s1_c4_uc = np.abs(m_s1_c4_uc_g - mi_s1_c4_uc_g)

abs_err_s2_c1_uc = np.abs(m_s2_c1_uc_g - mi_s2_c1_uc_g)
abs_err_s2_c2_uc = np.abs(m_s2_c2_uc_g - mi_s2_c2_uc_g)
abs_err_s2_c3_uc = np.abs(m_s2_c3_uc_g - mi_s2_c3_uc_g)
abs_err_s2_c4_uc = np.abs(m_s2_c4_uc_g - mi_s2_c4_uc_g)

abs_err_s3_c1_uc = np.abs(m_s3_c1_uc_g - mi_s3_c1_uc_g)
abs_err_s3_c2_uc = np.abs(m_s3_c2_uc_g - mi_s3_c2_uc_g)
abs_err_s3_c3_uc = np.abs(m_s3_c3_uc_g - mi_s3_c3_uc_g)
abs_err_s3_c4_uc = np.abs(m_s3_c4_uc_g - mi_s3_c4_uc_g)

abs_err_s4_c1_uc = np.abs(m_s4_c1_uc_g - mi_s4_c1_uc_g)
abs_err_s4_c2_uc = np.abs(m_s4_c2_uc_g - mi_s4_c2_uc_g)
abs_err_s4_c3_uc = np.abs(m_s4_c3_uc_g - mi_s4_c3_uc_g)
abs_err_s4_c4_uc = np.abs(m_s4_c4_uc_g - mi_s4_c4_uc_g)

```

```python
def PlotModels(mt_c1, mt_c2, mt_c3, mt_c4, mi_c1, mi_c2, mi_c3, mi_c4, rd_c1, rd_c2, rd_c3, rd_c4, slope=''):
    fig, ax = plt.subplots(4,3, sharex=True, sharey=True, figsize=(10,6))

    inputs = {'cMin':10,
              'cMax':2000,
              'colorBar' : True,
              'orientation' : 'vertical',
               # 'label': '$\sigma$ [mS/m]',
              'fontsize':7, 
              'labelsize':7
              }

    zMax=6
    fs = 7

    mt_c1[:,1:] = mt_c1[:,1:]*1000 # Transforming to miliSiemens
    mt_c2[:,1:] = mt_c2[:,1:]*1000
    mt_c3[:,1:] = mt_c3[:,1:]*1000
    mt_c4[:,1:] = mt_c4[:,1:]*1000

    mi_c1[:,1:] = mi_c1[:,1:]*1000 # Transforming to miliSiemens
    mi_c2[:,1:] = mi_c2[:,1:]*1000
    mi_c3[:,1:] = mi_c3[:,1:]*1000
    mi_c4[:,1:] = mi_c4[:,1:]*1000

    rd_c1[:,1:] = rd_c1[:,1:]*1000 # Transforming to miliSiemens
    rd_c2[:,1:] = rd_c2[:,1:]*1000
    rd_c3[:,1:] = rd_c3[:,1:]*1000
    rd_c4[:,1:] = rd_c4[:,1:]*1000

    RMSE_c1 = root_mean_squared_error(mt_c1, mi_c1)
    RMSE_c2 = root_mean_squared_error(mt_c2, mi_c2)
    RMSE_c3 = root_mean_squared_error(mt_c3, mi_c3)
    RMSE_c4 = root_mean_squared_error(mt_c4, mi_c4)
    
    # c1
    showStitchedModels(mt_c1, ax=ax[0,0], **inputs, zMax=zMax)
    ax[0,0].set_title('True model - Slope '+slope+'%', fontsize=fs)

    showStitchedModels(mi_c1, ax=ax[0,1], **inputs, zMax=zMax)
    ax[0,1].set_title('Inverted model - Slope '+slope+'%' + ' RMSE: %.2f' % RMSE_c1 + ' mS', fontsize=fs)

    c = ax[0,2].imshow(rd_c1.T, extent=[0,40,-6,0], aspect=3.6, cmap='binary', norm='log', vmin=1e-2, vmax=1e3)
    ax[0,2].set_title('Abs Error', fontsize=fs)
    cb = fig.colorbar(c, ax=ax[0,2])

    # c2
    showStitchedModels(mt_c2, ax=ax[1,0], **inputs, zMax=zMax)
    ax[1,0].set_title('True model - Slope '+slope+'%', fontsize=fs)

    showStitchedModels(mi_c2, ax=ax[1,1], **inputs, zMax=zMax)
    ax[1,1].set_title('Inverted model - Slope '+slope+'%'+ ' RMSE: %.2f' % RMSE_c2 + ' mS', fontsize=fs)

    c = ax[1,2].imshow(rd_c2.T, extent=[0,40,-6,0], aspect=3.6, cmap='binary', norm='log', vmin=1e-2, vmax=1e3)
    ax[1,2].set_title('Abs Error', fontsize=fs)
    cb = fig.colorbar(c, ax=ax[1,2])

    # c3
    showStitchedModels(mt_c3, ax=ax[2,0], **inputs, zMax=zMax)
    ax[2,0].set_title('True model - Slope '+slope+'%', fontsize=fs)

    showStitchedModels(mi_c3, ax=ax[2,1], **inputs, zMax=zMax)
    ax[2,1].set_title('Inverted model - Slope '+slope+'%'+ ' RMSE: %.2f' % RMSE_c3 + ' mS', fontsize=fs)

    c = ax[2,2].imshow(rd_c3.T, extent=[0,40,-6,0], aspect=3.6, cmap='binary', norm='log', vmin=1e-2, vmax=1e3)
    ax[2,2].set_title('Abs Error', fontsize=fs)
    cb = fig.colorbar(c, ax=ax[2,2])

    # c4
    showStitchedModels(mt_c4, ax=ax[3,0], **inputs, zMax=zMax)
    ax[3,0].set_title('True model - Slope '+slope+'%', fontsize=fs)

    showStitchedModels(mi_c4, ax=ax[3,1], **inputs, zMax=zMax)
    ax[3,1].set_title('Inverted model - Slope '+slope+'%'+ ' RMSE: %.2f' % RMSE_c4 + ' mS', fontsize=fs)

    c = ax[3,2].imshow(rd_c4.T, extent=[0,40,-6,0], aspect=3.6, cmap='binary', norm='log', vmin=1e-2, vmax=1e3)
    ax[3,2].set_title('Abs Error', fontsize=fs)
    cb = fig.colorbar(c, ax=ax[3,2])


    plt.tight_layout()
```

```python
PlotModels(model_s1_c1_uc, model_s1_c2_uc, model_s1_c3_uc, model_s1_c4_uc,
           model_inv_s1_c1_uc, model_inv_s1_c2_uc, model_inv_s1_c3_uc, model_inv_s1_c4_uc,
           abs_err_s1_c1_uc, abs_err_s1_c2_uc, abs_err_s1_c3_uc, abs_err_s1_c4_uc, slope='1')
```

```python
PlotModels(model_s3_c1_uc, model_s3_c2_uc, model_s3_c3_uc, model_s3_c4_uc,
           model_inv_s3_c1_uc, model_inv_s3_c2_uc, model_inv_s3_c3_uc, model_inv_s3_c4_uc,
           abs_err_s3_c1_uc, abs_err_s3_c2_uc, abs_err_s3_c3_uc, abs_err_s3_c4_uc, slope='5')
```

```python
PlotModels(model_s4_c1_uc, model_s4_c2_uc, model_s4_c3_uc, model_s4_c4_uc,
           model_inv_s4_c1_uc, model_inv_s4_c2_uc, model_inv_s4_c3_uc, model_inv_s4_c4_uc,
           abs_err_s4_c1_uc, abs_err_s4_c2_uc, abs_err_s4_c3_uc, abs_err_s4_c4_uc, slope='10')
```
