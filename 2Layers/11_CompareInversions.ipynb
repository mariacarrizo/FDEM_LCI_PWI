```python
def PlotComparison(mt_c1, mt_c2, mt_c3, mt_c4, 
                   mi_c1, mi_c2, mi_c3, mi_c4, 
                   mi_c1_pw, mi_c2_pw, mi_c3_pw, mi_c4_pw,
                   mi_c1_ns, mi_c2_ns, mi_c3_ns, mi_c4_ns,
                   mi_c1_pw_ns, mi_c2_pw_ns, mi_c3_pw_ns, mi_c4_pw_ns,
                   slope=''):
    
    fig, ax = plt.subplots(8,5, sharex=True, sharey=True, figsize=(12,10))

    inputs = {'cMin':10,
              'cMax':2000,
              'colorBar' : True,
              'orientation' : 'vertical',
               # 'label': '$\sigma$ [mS/m]',
              'fontsize':7, 
              'labelsize':7
              }

    zMax=6
    fs = 7
    asp=3.3
    
    mt_c1[:,1:] = mt_c1[:,1:]*1000 # Transforming to miliSiemens
    mt_c2[:,1:] = mt_c2[:,1:]*1000
    mt_c3[:,1:] = mt_c3[:,1:]*1000
    mt_c4[:,1:] = mt_c4[:,1:]*1000

    mi_c1[:,1:] = mi_c1[:,1:]*1000 # Transforming to miliSiemens
    mi_c2[:,1:] = mi_c2[:,1:]*1000
    mi_c3[:,1:] = mi_c3[:,1:]*1000
    mi_c4[:,1:] = mi_c4[:,1:]*1000

    mi_c1_pw[:,1:] = mi_c1_pw[:,1:]*1000 # Transforming to miliSiemens
    mi_c2_pw[:,1:] = mi_c2_pw[:,1:]*1000
    mi_c3_pw[:,1:] = mi_c3_pw[:,1:]*1000
    mi_c4_pw[:,1:] = mi_c4_pw[:,1:]*1000
    
    mi_c1_ns[:,1:] = mi_c1_ns[:,1:]*1000 # Transforming to miliSiemens
    mi_c2_ns[:,1:] = mi_c2_ns[:,1:]*1000
    mi_c3_ns[:,1:] = mi_c3_ns[:,1:]*1000
    mi_c4_ns[:,1:] = mi_c4_ns[:,1:]*1000

    mi_c1_pw_ns[:,1:] = mi_c1_pw_ns[:,1:]*1000 # Transforming to miliSiemens
    mi_c2_pw_ns[:,1:] = mi_c2_pw_ns[:,1:]*1000
    mi_c3_pw_ns[:,1:] = mi_c3_pw_ns[:,1:]*1000
    mi_c4_pw_ns[:,1:] = mi_c4_pw_ns[:,1:]*1000

    # Create grids
    mt_c1_g = grid(mt_c1, depthmax=6, nlay=2)
    mt_c2_g = grid(mt_c2, depthmax=6, nlay=2)
    mt_c3_g = grid(mt_c3, depthmax=6, nlay=2)
    mt_c4_g = grid(mt_c4, depthmax=6, nlay=2)

    mi_c1_g = grid(mi_c1, depthmax=6, nlay=2)
    mi_c2_g = grid(mi_c2, depthmax=6, nlay=2)
    mi_c3_g = grid(mi_c3, depthmax=6, nlay=2)
    mi_c4_g = grid(mi_c4, depthmax=6, nlay=2)

    mi_c1_pw_g = grid(mi_c1_pw, depthmax=6, nlay=2)
    mi_c2_pw_g = grid(mi_c2_pw, depthmax=6, nlay=2)
    mi_c3_pw_g = grid(mi_c3_pw, depthmax=6, nlay=2)
    mi_c4_pw_g = grid(mi_c4_pw, depthmax=6, nlay=2)

    mi_c1_ns_g = grid(mi_c1_ns, depthmax=6, nlay=2)
    mi_c2_ns_g = grid(mi_c2_ns, depthmax=6, nlay=2)
    mi_c3_ns_g = grid(mi_c3_ns, depthmax=6, nlay=2)
    mi_c4_ns_g = grid(mi_c4_ns, depthmax=6, nlay=2)

    mi_c1_pw_ns_g = grid(mi_c1_pw_ns, depthmax=6, nlay=2)
    mi_c2_pw_ns_g = grid(mi_c2_pw_ns, depthmax=6, nlay=2)
    mi_c3_pw_ns_g = grid(mi_c3_pw_ns, depthmax=6, nlay=2)
    mi_c4_pw_ns_g = grid(mi_c4_pw_ns, depthmax=6, nlay=2)

    # Calculate RMSE sigma

    RMSE_c1_s = root_mean_squared_error(mt_c1[:,1:], mi_c1[:,1:])
    RMSE_c2_s = root_mean_squared_error(mt_c2[:,1:], mi_c2[:,1:])
    RMSE_c3_s = root_mean_squared_error(mt_c3[:,1:], mi_c3[:,1:])
    RMSE_c4_s = root_mean_squared_error(mt_c4[:,1:], mi_c4[:,1:])

    RMSE_c1_pw_s = root_mean_squared_error(mt_c1[:,1:], mi_c1_pw[:,1:])
    RMSE_c2_pw_s = root_mean_squared_error(mt_c2[:,1:], mi_c2_pw[:,1:])
    RMSE_c3_pw_s = root_mean_squared_error(mt_c3[:,1:], mi_c3_pw[:,1:])
    RMSE_c4_pw_s = root_mean_squared_error(mt_c4[:,1:], mi_c4_pw[:,1:])
    
    RMSE_c1_ns_s = root_mean_squared_error(mt_c1[:,1:], mi_c1_ns[:,1:])
    RMSE_c2_ns_s = root_mean_squared_error(mt_c2[:,1:], mi_c2_ns[:,1:])
    RMSE_c3_ns_s = root_mean_squared_error(mt_c3[:,1:], mi_c3_ns[:,1:])
    RMSE_c4_ns_s = root_mean_squared_error(mt_c4[:,1:], mi_c4_ns[:,1:])

    RMSE_c1_pw_ns_s = root_mean_squared_error(mt_c1[:,1:], mi_c1_pw_ns[:,1:])
    RMSE_c2_pw_ns_s = root_mean_squared_error(mt_c2[:,1:], mi_c2_pw_ns[:,1:])
    RMSE_c3_pw_ns_s = root_mean_squared_error(mt_c3[:,1:], mi_c3_pw_ns[:,1:])
    RMSE_c4_pw_ns_s = root_mean_squared_error(mt_c4[:,1:], mi_c4_pw_ns[:,1:])

    # Calculate RMSE thickness

    RMSE_c1_h = root_mean_squared_error(mt_c1[:,0], mi_c1[:,0])
    RMSE_c2_h = root_mean_squared_error(mt_c2[:,0], mi_c2[:,0])
    RMSE_c3_h = root_mean_squared_error(mt_c3[:,0], mi_c3[:,0])
    RMSE_c4_h = root_mean_squared_error(mt_c4[:,0], mi_c4[:,0])

    RMSE_c1_pw_h = root_mean_squared_error(mt_c1[:,0], mi_c1_pw[:,0])
    RMSE_c2_pw_h = root_mean_squared_error(mt_c2[:,0], mi_c2_pw[:,0])
    RMSE_c3_pw_h = root_mean_squared_error(mt_c3[:,0], mi_c3_pw[:,0])
    RMSE_c4_pw_h = root_mean_squared_error(mt_c4[:,0], mi_c4_pw[:,0])

    RMSE_c1_ns_h = root_mean_squared_error(mt_c1[:,0], mi_c1_ns[:,0])
    RMSE_c2_ns_h = root_mean_squared_error(mt_c2[:,0], mi_c2_ns[:,0])
    RMSE_c3_ns_h = root_mean_squared_error(mt_c3[:,0], mi_c3_ns[:,0])
    RMSE_c4_ns_h = root_mean_squared_error(mt_c4[:,0], mi_c4_ns[:,0])

    RMSE_c1_pw_ns_h = root_mean_squared_error(mt_c1[:,0], mi_c1_pw_ns[:,0])
    RMSE_c2_pw_ns_h = root_mean_squared_error(mt_c2[:,0], mi_c2_pw_ns[:,0])
    RMSE_c3_pw_ns_h = root_mean_squared_error(mt_c3[:,0], mi_c3_pw_ns[:,0])
    RMSE_c4_pw_ns_h = root_mean_squared_error(mt_c4[:,0], mi_c4_pw_ns[:,0])

    # Calculate absolute errors

    ae_c1 = np.abs(mt_c1_g - mi_c1_g)
    ae_c2 = np.abs(mt_c2_g - mi_c2_g)
    ae_c3 = np.abs(mt_c3_g - mi_c3_g)
    ae_c4 = np.abs(mt_c4_g - mi_c4_g)

    ae_c1_pw = np.abs(mt_c1_g - mi_c1_pw_g)
    ae_c2_pw = np.abs(mt_c2_g - mi_c2_pw_g)
    ae_c3_pw = np.abs(mt_c3_g - mi_c3_pw_g)
    ae_c4_pw = np.abs(mt_c4_g - mi_c4_pw_g)

    ae_c1_ns = np.abs(mt_c1_g - mi_c1_ns_g)
    ae_c2_ns = np.abs(mt_c2_g - mi_c2_ns_g)
    ae_c3_ns = np.abs(mt_c3_g - mi_c3_ns_g)
    ae_c4_ns = np.abs(mt_c4_g - mi_c4_ns_g)

    ae_c1_pw_ns = np.abs(mt_c1_g - mi_c1_pw_ns_g)
    ae_c2_pw_ns = np.abs(mt_c2_g - mi_c2_pw_ns_g)
    ae_c3_pw_ns = np.abs(mt_c3_g - mi_c3_pw_ns_g)
    ae_c4_pw_ns = np.abs(mt_c4_g - mi_c4_pw_ns_g)
    
    # Results c1
    showStitchedModels(mt_c1, ax=ax[0,0], **inputs, zMax=zMax)
    ax[0,0].set_title('True model', fontsize=fs)

    showStitchedModels(mi_c1, ax=ax[0,1], **inputs, zMax=zMax)
    ax[0,1].set_title('LCI -' + ' RMSE $\sigma$: %.2f' % RMSE_c1_s + ' mS - $h$: %.2f' % RMSE_c1_h+' m', fontsize=fs)

    showStitchedModels(mi_c1_ns, ax=ax[0,2], **inputs, zMax=zMax)
    ax[0,2].set_title('LCI no sort -' + ' RMSE $\sigma$: %.2f' % RMSE_c1_ns_s + ' mS - $h$: %.2f' % RMSE_c1_ns_h+' m', fontsize=fs)

    showStitchedModels(mi_c1_pw, ax=ax[0,3], **inputs, zMax=zMax)
    ax[0,3].set_title('Piecewise -' + ' RMSE $\sigma$: %.2f' % RMSE_c1_pw_s + ' mS - $h$: %.2f' % RMSE_c1_pw_h+' m', fontsize=fs)

    showStitchedModels(mi_c1_pw_ns, ax=ax[0,4], **inputs, zMax=zMax)
    ax[0,4].set_title('PW no sort -' + ' RMSE $\sigma$: %.2f' % RMSE_c1_pw_s + ' mS - $h$: %.2f' % RMSE_c1_pw_ns_h+' m', fontsize=fs)

    # Absolute error c1
    
    c = ax[1,1].imshow(ae_c1.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1e-2, vmax=1e3)
    ax[1,1].set_title('Abs Error', fontsize=fs)
    cb = fig.colorbar(c, ax=ax[1,1])

    c = ax[1,2].imshow(ae_c1_ns.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1e-2, vmax=1e3)
    ax[1,2].set_title('Abs Error', fontsize=fs)
    cb = fig.colorbar(c, ax=ax[1,2])

    c = ax[1,3].imshow(ae_c1_pw.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1e-2, vmax=1e3)
    ax[1,3].set_title('Abs Error', fontsize=fs)
    cb = fig.colorbar(c, ax=ax[1,3])

    c = ax[1,4].imshow(ae_c1_pw_ns.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1e-2, vmax=1e3)
    ax[1,4].set_title('Abs Error', fontsize=fs)
    cb = fig.colorbar(c, ax=ax[1,4])

    # c2
    showStitchedModels(mt_c2, ax=ax[2,0], **inputs, zMax=zMax)
    ax[2,0].set_title('True model ', fontsize=fs)

    showStitchedModels(mi_c2, ax=ax[2,1], **inputs, zMax=zMax)
    ax[2,1].set_title('LCI -' + ' RMSE $\sigma$: %.2f' % RMSE_c2_s + ' mS - $h$: %.2f' % RMSE_c2_h+' m', fontsize=fs)

    showStitchedModels(mi_c2_ns, ax=ax[2,2], **inputs, zMax=zMax)
    ax[2,2].set_title('LCI no sort -' + ' RMSE $\sigma$: %.2f' % RMSE_c2_ns_s + ' mS - $h$: %.2f' % RMSE_c2_ns_h+' m', fontsize=fs)

    showStitchedModels(mi_c2_pw, ax=ax[2,3], **inputs, zMax=zMax)
    ax[2,3].set_title('Piecewise -' + ' RMSE $\sigma$: %.2f' % RMSE_c2_pw_s + ' mS - $h$: %.2f' % RMSE_c2_pw_h+' m', fontsize=fs)

    showStitchedModels(mi_c2_pw_ns, ax=ax[2,4], **inputs, zMax=zMax)
    ax[2,4].set_title('PW no sort -' + ' RMSE $\sigma$: %.2f' % RMSE_c2_pw_ns_s + ' mS - $h$: %.2f' % RMSE_c2_pw_ns_h+' m', fontsize=fs)

    # Absolute error c2
    
    c = ax[3,1].imshow(ae_c2.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1e-2, vmax=1e3)
    ax[3,1].set_title('Abs Error', fontsize=fs)
    cb = fig.colorbar(c, ax=ax[3,1])

    c = ax[3,2].imshow(ae_c2_ns.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1e-2, vmax=1e3)
    ax[3,2].set_title('Abs Error', fontsize=fs)
    cb = fig.colorbar(c, ax=ax[3,2])

    c = ax[3,3].imshow(ae_c2_pw.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1e-2, vmax=1e3)
    ax[3,3].set_title('Abs Error', fontsize=fs)
    cb = fig.colorbar(c, ax=ax[3,3])

    c = ax[3,4].imshow(ae_c2_pw_ns.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1e-2, vmax=1e3)
    ax[3,4].set_title('Abs Error', fontsize=fs)
    cb = fig.colorbar(c, ax=ax[3,4])

    # c3
    showStitchedModels(mt_c3, ax=ax[4,0], **inputs, zMax=zMax)
    ax[4,0].set_title('True model ', fontsize=fs)

    showStitchedModels(mi_c3, ax=ax[4,1], **inputs, zMax=zMax)
    ax[4,1].set_title('LCI -' + ' RMSE $\sigma$: %.2f' % RMSE_c3_s + ' mS - $h$: %.2f' % RMSE_c3_h+' m', fontsize=fs)

    showStitchedModels(mi_c3_ns, ax=ax[4,2], **inputs, zMax=zMax)
    ax[4,2].set_title('LCI no sort -' + ' RMSE $\sigma$: %.2f' % RMSE_c3_ns_s + ' mS - $h$: %.2f' % RMSE_c3_ns_h+' m', fontsize=fs)

    showStitchedModels(mi_c3_pw, ax=ax[4,3], **inputs, zMax=zMax)
    ax[4,3].set_title('Piecewise -' + ' RMSE $\sigma$: %.2f' % RMSE_c3_pw_s + ' mS - $h$: %.2f' % RMSE_c3_pw_h+' m', fontsize=fs)

    showStitchedModels(mi_c3_pw_ns, ax=ax[4,4], **inputs, zMax=zMax)
    ax[4,4].set_title('PW no sort -' + ' RMSE $\sigma$: %.2f' % RMSE_c3_pw_ns_s + ' mS - $h$: %.2f' % RMSE_c3_pw_ns_h+' m', fontsize=fs)

    # Absolute error c3
    
    c = ax[5,1].imshow(ae_c3.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1e-2, vmax=1e3)
    ax[5,1].set_title('Abs Error', fontsize=fs)
    cb = fig.colorbar(c, ax=ax[5,1])

    c = ax[5,2].imshow(ae_c3_ns.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1e-2, vmax=1e3)
    ax[5,2].set_title('Abs Error', fontsize=fs)
    cb = fig.colorbar(c, ax=ax[5,2])

    c = ax[5,3].imshow(ae_c3_pw.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1e-2, vmax=1e3)
    ax[5,3].set_title('Abs Error', fontsize=fs)
    cb = fig.colorbar(c, ax=ax[5,3])

    c = ax[5,4].imshow(ae_c3_pw_ns.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1e-2, vmax=1e3)
    ax[5,4].set_title('Abs Error', fontsize=fs)
    cb = fig.colorbar(c, ax=ax[5,4])

    # c4
    showStitchedModels(mt_c4, ax=ax[6,0], **inputs, zMax=zMax)
    ax[6,0].set_title('True model ', fontsize=fs)

    showStitchedModels(mi_c4, ax=ax[6,1], **inputs, zMax=zMax)
    ax[6,1].set_title('LCI -' + ' RMSE $\sigma$: %.2f' % RMSE_c4_s + ' mS - $h$: %.2f' % RMSE_c4_h+' m', fontsize=fs)

    showStitchedModels(mi_c4_ns, ax=ax[6,2], **inputs, zMax=zMax)
    ax[6,2].set_title('LCI no sort -' + ' RMSE $\sigma$: %.2f' % RMSE_c4_ns_s + ' mS - $h$: %.2f' % RMSE_c4_ns_h+' m', fontsize=fs)

    showStitchedModels(mi_c4_pw, ax=ax[6,3], **inputs, zMax=zMax)
    ax[6,3].set_title('Piecewise -' + ' RMSE $\sigma$: %.2f' % RMSE_c4_pw_s + ' mS - $h$: %.2f' % RMSE_c4_pw_h+' m', fontsize=fs)

    showStitchedModels(mi_c4_pw_ns, ax=ax[6,4], **inputs, zMax=zMax)
    ax[6,4].set_title('PW no sort -' + ' RMSE $\sigma$: %.2f' % RMSE_c4_pw_ns_s + ' mS - $h$: %.2f' % RMSE_c4_pw_ns_h+' m', fontsize=fs)

    # Absolute error c4
    
    c = ax[7,1].imshow(ae_c4.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1e-2, vmax=1e3)
    ax[7,1].set_title('Abs Error', fontsize=fs)
    cb = fig.colorbar(c, ax=ax[7,1])

    c = ax[7,2].imshow(ae_c4_ns.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1e-2, vmax=1e3)
    ax[7,2].set_title('Abs Error', fontsize=fs)
    cb = fig.colorbar(c, ax=ax[7,2])

    c = ax[7,3].imshow(ae_c4_pw.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1e-2, vmax=1e3)
    ax[7,3].set_title('Abs Error', fontsize=fs)
    cb = fig.colorbar(c, ax=ax[7,3])

    c = ax[7,4].imshow(ae_c4_pw_ns.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1e-2, vmax=1e3)
    ax[7,4].set_title('Abs Error', fontsize=fs)
    cb = fig.colorbar(c, ax=ax[7,4])

    fig.delaxes(ax[1,0])
    fig.delaxes(ax[3,0])
    fig.delaxes(ax[5,0])
    fig.delaxes(ax[7,0])

    fig.suptitle('Slope: '+slope+'%')

    plt.tight_layout()
```

```python
def PlotComparison2(mt_c1, mt_c2, mt_c3, mt_c4, 
                   mi_c1, mi_c2, mi_c3, mi_c4, 
                   mi_c1_pw, mi_c2_pw, mi_c3_pw, mi_c4_pw,
                   slope=''):
    
    fig, ax = plt.subplots(8,3, sharex=True, sharey=True, figsize=(12,10))

    inputs = {'cMin':10,
              'cMax':2000,
              'colorBar' : True,
              'orientation' : 'vertical',
              'label': '$\sigma$ [mS/m]',
              'fontsize':7, 
              'labelsize':7
              }

    zMax=6
    fs = 7
    asp= 1.75
    
    mt_c1[:,1:] = mt_c1[:,1:]*1000 # Transforming to miliSiemens
    mt_c2[:,1:] = mt_c2[:,1:]*1000
    mt_c3[:,1:] = mt_c3[:,1:]*1000
    mt_c4[:,1:] = mt_c4[:,1:]*1000

    mi_c1[:,1:] = mi_c1[:,1:]*1000 # Transforming to miliSiemens
    mi_c2[:,1:] = mi_c2[:,1:]*1000
    mi_c3[:,1:] = mi_c3[:,1:]*1000
    mi_c4[:,1:] = mi_c4[:,1:]*1000

    mi_c1_pw[:,1:] = mi_c1_pw[:,1:]*1000 # Transforming to miliSiemens
    mi_c2_pw[:,1:] = mi_c2_pw[:,1:]*1000
    mi_c3_pw[:,1:] = mi_c3_pw[:,1:]*1000
    mi_c4_pw[:,1:] = mi_c4_pw[:,1:]*1000

    # Create grids
    mt_c1_g = grid(mt_c1, depthmax=6, nlay=2)
    mt_c2_g = grid(mt_c2, depthmax=6, nlay=2)
    mt_c3_g = grid(mt_c3, depthmax=6, nlay=2)
    mt_c4_g = grid(mt_c4, depthmax=6, nlay=2)

    mi_c1_g = grid(mi_c1, depthmax=6, nlay=2)
    mi_c2_g = grid(mi_c2, depthmax=6, nlay=2)
    mi_c3_g = grid(mi_c3, depthmax=6, nlay=2)
    mi_c4_g = grid(mi_c4, depthmax=6, nlay=2)

    mi_c1_pw_g = grid(mi_c1_pw, depthmax=6, nlay=2)
    mi_c2_pw_g = grid(mi_c2_pw, depthmax=6, nlay=2)
    mi_c3_pw_g = grid(mi_c3_pw, depthmax=6, nlay=2)
    mi_c4_pw_g = grid(mi_c4_pw, depthmax=6, nlay=2)

    # Calculate RMSE sigma

    RMSE_c1_s = root_mean_squared_error(mt_c1[:,1:], mi_c1[:,1:])
    RMSE_c2_s = root_mean_squared_error(mt_c2[:,1:], mi_c2[:,1:])
    RMSE_c3_s = root_mean_squared_error(mt_c3[:,1:], mi_c3[:,1:])
    RMSE_c4_s = root_mean_squared_error(mt_c4[:,1:], mi_c4[:,1:])

    RMSE_c1_pw_s = root_mean_squared_error(mt_c1[:,1:], mi_c1_pw[:,1:])
    RMSE_c2_pw_s = root_mean_squared_error(mt_c2[:,1:], mi_c2_pw[:,1:])
    RMSE_c3_pw_s = root_mean_squared_error(mt_c3[:,1:], mi_c3_pw[:,1:])
    RMSE_c4_pw_s = root_mean_squared_error(mt_c4[:,1:], mi_c4_pw[:,1:])
    

    # Calculate RMSE thickness

    RMSE_c1_h = root_mean_squared_error(mt_c1[:,0], mi_c1[:,0])
    RMSE_c2_h = root_mean_squared_error(mt_c2[:,0], mi_c2[:,0])
    RMSE_c3_h = root_mean_squared_error(mt_c3[:,0], mi_c3[:,0])
    RMSE_c4_h = root_mean_squared_error(mt_c4[:,0], mi_c4[:,0])

    RMSE_c1_pw_h = root_mean_squared_error(mt_c1[:,0], mi_c1_pw[:,0])
    RMSE_c2_pw_h = root_mean_squared_error(mt_c2[:,0], mi_c2_pw[:,0])
    RMSE_c3_pw_h = root_mean_squared_error(mt_c3[:,0], mi_c3_pw[:,0])
    RMSE_c4_pw_h = root_mean_squared_error(mt_c4[:,0], mi_c4_pw[:,0])

    # Calculate absolute errors

    ae_c1 = np.abs(mt_c1_g - mi_c1_g)
    ae_c2 = np.abs(mt_c2_g - mi_c2_g)
    ae_c3 = np.abs(mt_c3_g - mi_c3_g)
    ae_c4 = np.abs(mt_c4_g - mi_c4_g)

    ae_c1_pw = np.abs(mt_c1_g - mi_c1_pw_g)
    ae_c2_pw = np.abs(mt_c2_g - mi_c2_pw_g)
    ae_c3_pw = np.abs(mt_c3_g - mi_c3_pw_g)
    ae_c4_pw = np.abs(mt_c4_g - mi_c4_pw_g)
    
    # Results c1
    showStitchedModels(mt_c1, ax=ax[0,0], **inputs, zMax=zMax)
    ax[0,0].set_title('True model', fontsize=fs)
    ax[0,0].tick_params(labelsize=fs)

    showStitchedModels(mi_c1, ax=ax[0,1], **inputs, zMax=zMax)
    ax[0,1].set_title('LCI -' + ' RMSE $\sigma$: %.2f' % RMSE_c1_s + ' mS - $h$: %.2f' % RMSE_c1_h+' m', fontsize=fs)
    ax[0,1].tick_params(labelsize=fs)
    ax[0,1].plot(-mt_c1[:,0], ':k')

    showStitchedModels(mi_c1_pw, ax=ax[0,2], **inputs, zMax=zMax)
    ax[0,2].set_title('Piecewise -' + ' RMSE $\sigma$: %.2f' % RMSE_c1_pw_s + ' mS - $h$: %.2f' % RMSE_c1_pw_h+' m', fontsize=fs)
    ax[0,2].tick_params(labelsize=fs)
    ax[0,2].plot(-mt_c1[:,0], ':k')


    # Absolute error c1
    
    c = ax[1,1].imshow(ae_c1.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1, vmax=1e2)
    ax[1,1].set_title('Abs Error', fontsize=fs)
    ax[1,1].tick_params(labelsize=fs)
    cb = fig.colorbar(c, ax=ax[1,1])
    cb.ax.tick_params(labelsize=fs)
    cb.set_label(label = '$\sigma$ [mS/m]', fontsize=fs)

    c = ax[1,2].imshow(ae_c1_pw.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1, vmax=1e2)
    ax[1,2].set_title('Abs Error', fontsize=fs)
    ax[1,2].tick_params(labelsize=fs)
    cb = fig.colorbar(c, ax=ax[1,2])
    cb.ax.tick_params(labelsize=fs)
    cb.set_label(label = '$\sigma$ [mS/m]', fontsize=fs)


    # c2
    showStitchedModels(mt_c2, ax=ax[2,0], **inputs, zMax=zMax)
    ax[2,0].set_title('True model ', fontsize=fs)
    ax[2,0].tick_params(labelsize=fs)

    showStitchedModels(mi_c2, ax=ax[2,1], **inputs, zMax=zMax)
    ax[2,1].set_title('LCI -' + ' RMSE $\sigma$: %.2f' % RMSE_c2_s + ' mS - $h$: %.2f' % RMSE_c2_h+' m', fontsize=fs)
    ax[2,1].tick_params(labelsize=fs)
    ax[2,1].plot(-mt_c2[:,0], ':k')

    showStitchedModels(mi_c2_pw, ax=ax[2,2], **inputs, zMax=zMax)
    ax[2,2].set_title('Piecewise -' + ' RMSE $\sigma$: %.2f' % RMSE_c2_pw_s + ' mS - $h$: %.2f' % RMSE_c2_pw_h+' m', fontsize=fs)
    ax[2,2].tick_params(labelsize=fs)
    ax[2,1].plot(-mt_c2[:,0], ':k')
    

    # Absolute error c2
    
    c = ax[3,1].imshow(ae_c2.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1, vmax=1e2)
    ax[3,1].set_title('Abs Error', fontsize=fs)
    ax[3,1].tick_params(labelsize=fs)
    cb = fig.colorbar(c, ax=ax[3,1])
    cb.ax.tick_params(labelsize=fs)
    cb.set_label(label = '$\sigma$ [mS/m]', fontsize=fs)

    c = ax[3,2].imshow(ae_c2_pw.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1, vmax=1e2)
    ax[3,2].set_title('Abs Error', fontsize=fs)
    ax[3,2].tick_params(labelsize=fs)
    cb = fig.colorbar(c, ax=ax[3,2])
    cb.ax.tick_params(labelsize=fs)
    cb.set_label(label = '$\sigma$ [mS/m]', fontsize=fs)

    
    # c3
    showStitchedModels(mt_c3, ax=ax[4,0], **inputs, zMax=zMax)
    ax[4,0].set_title('True model ', fontsize=fs)
    ax[4,0].tick_params(labelsize=fs)
   
    showStitchedModels(mi_c3, ax=ax[4,1], **inputs, zMax=zMax)
    ax[4,1].set_title('LCI -' + ' RMSE $\sigma$: %.2f' % RMSE_c3_s + ' mS - $h$: %.2f' % RMSE_c3_h+' m', fontsize=fs)
    ax[4,1].tick_params(labelsize=fs)
    ax[4,1].plot(-mt_c3[:,0], ':k')

    showStitchedModels(mi_c3_pw, ax=ax[4,2], **inputs, zMax=zMax)
    ax[4,2].set_title('Piecewise -' + ' RMSE $\sigma$: %.2f' % RMSE_c3_pw_s + ' mS - $h$: %.2f' % RMSE_c3_pw_h+' m', fontsize=fs)
    ax[4,2].tick_params(labelsize=fs)
    ax[4,2].plot(-mt_c3[:,0], ':k')


    # Absolute error c3
    
    c = ax[5,1].imshow(ae_c3.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1, vmax=1e2)
    ax[5,1].set_title('Abs Error', fontsize=fs)
    ax[5,1].tick_params(labelsize=fs)
    cb = fig.colorbar(c, ax=ax[5,1])
    cb.ax.tick_params(labelsize=fs)
    cb.set_label(label = '$\sigma$ [mS/m]', fontsize=fs)

    c = ax[5,2].imshow(ae_c3_pw.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1, vmax=1e2)
    ax[5,2].set_title('Abs Error', fontsize=fs)
    ax[5,2].tick_params(labelsize=fs)
    cb = fig.colorbar(c, ax=ax[5,2])
    cb.ax.tick_params(labelsize=fs)
    cb.set_label(label = '$\sigma$ [mS/m]', fontsize=fs)


    # c4
    showStitchedModels(mt_c4, ax=ax[6,0], **inputs, zMax=zMax)
    ax[6,0].set_title('True model ', fontsize=fs)
    ax[6,0].tick_params(labelsize=fs)

    showStitchedModels(mi_c4, ax=ax[6,1], **inputs, zMax=zMax)
    ax[6,1].set_title('LCI -' + ' RMSE $\sigma$: %.2f' % RMSE_c4_s + ' mS - $h$: %.2f' % RMSE_c4_h+' m', fontsize=fs)
    ax[6,1].tick_params(labelsize=fs)
    ax[6,1].plot(-mt_c4[:,0], ':k')

    showStitchedModels(mi_c4_pw, ax=ax[6,2], **inputs, zMax=zMax)
    ax[6,2].set_title('Piecewise -' + ' RMSE $\sigma$: %.2f' % RMSE_c4_pw_s + ' mS - $h$: %.2f' % RMSE_c4_pw_h+' m', fontsize=fs)
    ax[6,2].tick_params(labelsize=fs)
    ax[6,2].plot(-mt_c4[:,0], ':k')
    

    # Absolute error c4
    
    c = ax[7,1].imshow(ae_c4.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1, vmax=1e2)
    ax[7,1].set_title('Abs Error', fontsize=fs)
    ax[7,1].tick_params(labelsize=fs)
    cb = fig.colorbar(c, ax=ax[7,1])
    cb.ax.tick_params(labelsize=fs)
    cb.set_label(label = '$\sigma$ [mS/m]', fontsize=fs)

    c = ax[7,2].imshow(ae_c4_pw.T, extent=[0,40,-6,0], aspect=asp, cmap='binary', norm='log', vmin=1, vmax=1e2)
    ax[7,2].set_title('Abs Error', fontsize=fs)
    ax[7,2].tick_params(labelsize=fs)
    cb = fig.colorbar(c, ax=ax[7,2])
    cb.ax.tick_params(labelsize=fs)
    cb.set_label(label = '$\sigma$ [mS/m]', fontsize=fs)


    fig.delaxes(ax[1,0])
    fig.delaxes(ax[3,0])
    fig.delaxes(ax[5,0])
    fig.delaxes(ax[7,0])

    fig.suptitle('Slope: '+slope+'%')

    plt.tight_layout()
```
