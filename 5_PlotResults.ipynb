```python
# Calculate relative error

rel_diff_s1_c1 = np.log10(np.abs((m_s1_c1_uc_g-mi_s1_c1_uc_g)/m_s1_c1_uc_g))
rel_diff_s1_c2 = np.log10(np.abs((m_s1_c2_uc_g-mi_s1_c2_uc_g)/m_s1_c2_uc_g))
rel_diff_s1_c3 = np.log10(np.abs((m_s1_c3_uc_g-mi_s1_c3_uc_g)/m_s1_c3_uc_g))
rel_diff_s1_c4 = np.log10(np.abs((m_s1_c4_uc_g-mi_s1_c4_uc_g)/m_s1_c4_uc_g))

rel_diff_s1_c1_ur = np.log10(np.abs((m_s1_c1_ur_g-mi_s1_c1_ur_g)/m_s1_c1_ur_g))
rel_diff_s1_c2_ur = np.log10(np.abs((m_s1_c2_ur_g-mi_s1_c2_ur_g)/m_s1_c2_ur_g))
rel_diff_s1_c3_ur = np.log10(np.abs((m_s1_c3_ur_g-mi_s1_c3_ur_g)/m_s1_c3_ur_g))
rel_diff_s1_c4_ur = np.log10(np.abs((m_s1_c4_ur_g-mi_s1_c4_ur_g)/m_s1_c4_ur_g))

rel_diff_s2_c1 = np.log10(np.abs((m_s2_c1_uc_g-mi_s2_c1_uc_g)/m_s2_c1_uc_g))
rel_diff_s2_c2 = np.log10(np.abs((m_s2_c2_uc_g-mi_s2_c2_uc_g)/m_s2_c2_uc_g))
rel_diff_s2_c3 = np.log10(np.abs((m_s2_c3_uc_g-mi_s2_c3_uc_g)/m_s2_c3_uc_g))
rel_diff_s2_c4 = np.log10(np.abs((m_s2_c4_uc_g-mi_s2_c4_uc_g)/m_s2_c4_uc_g))

rel_diff_s3_c1 = np.log10(np.abs((m_s3_c1_uc_g-mi_s3_c1_uc_g)/m_s3_c1_uc_g))
rel_diff_s3_c2 = np.log10(np.abs((m_s3_c2_uc_g-mi_s3_c2_uc_g)/m_s3_c2_uc_g))
rel_diff_s3_c3 = np.log10(np.abs((m_s3_c3_uc_g-mi_s3_c3_uc_g)/m_s3_c3_uc_g))
rel_diff_s3_c4 = np.log10(np.abs((m_s3_c4_uc_g-mi_s3_c4_uc_g)/m_s3_c4_uc_g))

rel_diff_s4_c1 = np.log10(np.abs((m_s4_c1_uc_g-mi_s4_c1_uc_g)/m_s4_c1_uc_g))
rel_diff_s4_c2 = np.log10(np.abs((m_s4_c2_uc_g-mi_s4_c2_uc_g)/m_s4_c2_uc_g))
rel_diff_s4_c3 = np.log10(np.abs((m_s4_c3_uc_g-mi_s4_c3_uc_g)/m_s4_c3_uc_g))
rel_diff_s4_c4 = np.log10(np.abs((m_s4_c4_uc_g-mi_s4_c4_uc_g)/m_s4_c4_uc_g))
```

```python
def PlotModels(mt_c1, mt_c2, mt_c3, mt_c4, mi_c1, mi_c2, mi_c3, mi_c4, rd_c1, rd_c2, rd_c3, rd_c4, slope=''):
    fig, ax = plt.subplots(4,3, sharex=True, sharey=True, figsize=(10,6))

    inputs = {'cMin':10/1000,
              'cMax':1800/1000,
              'colorBar' : True,
              'orientation' : 'vertical',
               # 'label': '$\sigma$ [mS/m]',
              'fontsize':7, 
              'labelsize':7
              }

    zMax=6
    fs = 7

    # c1
    showStitchedModels(mt_c1, ax=ax[0,0], **inputs, zMax=zMax)
    ax[0,0].set_title('True model - Slope '+slope+'%', fontsize=fs)

    showStitchedModels(mi_c1, ax=ax[0,1], **inputs, zMax=zMax)
    ax[0,1].set_title('Inverted model - Slope '+slope+'%', fontsize=fs)

    c = ax[0,2].imshow(rd_c1.T, extent=[0,40,-6,0], aspect=3.6, cmap='binary')
    ax[0,2].set_title('Log10 (Rel diff)', fontsize=fs)
    cb = fig.colorbar(c, ax=ax[0,2])

    # c2
    showStitchedModels(mt_c2, ax=ax[1,0], **inputs, zMax=zMax)
    ax[1,0].set_title('True model - Slope '+slope+'%', fontsize=fs)

    showStitchedModels(mi_c2, ax=ax[1,1], **inputs, zMax=zMax)
    ax[1,1].set_title('Inverted model - Slope '+slope+'%', fontsize=fs)

    ax[1,2].imshow(rd_c2.T, extent=[0,40,-6,0], aspect=3.6, cmap='binary')
    ax[1,2].set_title('Log10 (Rel diff)', fontsize=fs)

    # c3
    showStitchedModels(mt_c3, ax=ax[2,0], **inputs, zMax=zMax)
    ax[2,0].set_title('True model - Slope '+slope+'%', fontsize=fs)

    showStitchedModels(mi_c3, ax=ax[2,1], **inputs, zMax=zMax)
    ax[2,1].set_title('Inverted model - Slope '+slope+'%', fontsize=fs)

    ax[2,2].imshow(rd_c3.T, extent=[0,40,-6,0], aspect=3.6, cmap='binary')
    ax[2,2].set_title('Log10 (Rel diff)', fontsize=fs)

    # c4
    showStitchedModels(mt_c4, ax=ax[3,0], **inputs, zMax=zMax)
    ax[3,0].set_title('True model - Slope '+slope+'%', fontsize=fs)

    showStitchedModels(mi_c4, ax=ax[3,1], **inputs, zMax=zMax)
    ax[3,1].set_title('Inverted model - Slope '+slope+'%', fontsize=fs)

    ax[3,2].imshow(rd_c4.T, extent=[0,40,-6,0], aspect=3.6, cmap='binary')
    ax[3,2].set_title('Log10 (Rel diff)', fontsize=fs)


    plt.tight_layout()
```

```python
PlotModels(model_s1_c1_ur, model_s1_c2_ur, model_s1_c3_ur, model_s1_c4_ur,
           model_inv_s1_c1_ur, model_inv_s1_c2_ur, model_inv_s1_c3_ur, model_inv_s1_c4_ur,
           rel_diff_s1_c1_ur, rel_diff_s1_c2_ur, rel_diff_s1_c3_ur, rel_diff_s1_c4_ur, slope='1')
```
